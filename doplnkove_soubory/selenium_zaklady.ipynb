# Úvod do automatizace webu pomocí frameworku Selenium

Selenium WebDriver je framework pro automatizované testování a ovládání webových prohlížečů. Umožňuje nám programově simulovat chování běžného uživatele – otevírání stránek, klikání na elementy, vyplňování formulářů a extrakci dat z DOM (Document Object Model) struktury. Některé stránky mají ale ochranu před boty, takže na to pozor.

## 1. Příprava prostředí a závislostí
Pro běh Selenia potřebujeme nainstalovat samotnou knihovnu a nástroj pro správu ovladačů (driverů) prohlížeče. Driver funguje jako prostředník (proxy) mezi naším Python kódem a nativním API (ovládacím rozhraním) samotného prohlížeče.

## 2. Inicializace instance prohlížeče
V tomto kroku nastartujeme službu ChromeDriveru a vytvoříme instanci prohlížeče. Následně pomocí metody `get()` odešleme HTTP GET požadavek na cílovou URL adresu. V ukázce si zobrazíme mou osobní stránku.

```python
import time

from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.chrome.options import Options

# Initialize the ChromeDriver service
service = Service(ChromeDriverManager().install())

options = Options()

# Start the session
driver = webdriver.Chrome(options=options)

# Navigate to a specific URL
driver.get("https://askoldh.cz")

# Hard wait to observe the execution
time.sleep(3) # Uncomment this after first run

# Terminate the session and close browser processes
driver.quit()
```

## 3. Přehled základních příkazů v Seleniu

| Příkaz | Co to dělá                                                                                              |
| :--- |:--------------------------------------------------------------------------------------------------------|
| `driver.get("url")` | Otevře prohlížeč a načte zadanou webovou adresu.                                                        |
| `find_element(By.ID, "id")` | Najde na stránce prvek podle jeho přesného `id` (nejrychlejší a nejspolehlivější způsob).               |
| `find_element(By.NAME, "name")` | Najde prvek podle jeho jména (`name` atribut), často se tak hledají políčka pro vyplňování.             |
| `find_element(By.CSS_SELECTOR, ".trida")` | Hledá prvky pomocí CSS selektorů (stejně jako když v CSS styluješ stránku, např. podle tříd nebo ID).   |
| `find_element(By.XPATH, "//div")` | Hledá prvek složitější cestou přímo ve struktuře HTML kódu (když nejde použít ID ani třída).            |
| `element.send_keys("text")` | Napíše zadaný text do políčka, jako by to udělal člověk na klávesnici.                                  |
| `element.click()` | Simuluje kliknutí levým tlačítkem myši na vybraný prvek (tlačítko, odkaz).                              |
| `element.clear()` | Smaže všechen text, který už v políčku je (hodí se před psaním nového textu).                           |
| `driver.implicitly_wait(time)` | Řekne Seleniu, ať při hledání prvků chvíli počká (max zadaný počet vteřin), než to vzdá a vyhodí chybu. |
| `driver.title` | Získá text, který vidíš na záložce prohlížeče (název aktuální stránky).                                 |
| `driver.quit()` | Všechno korektně zavře – okno prohlížeče i proces ovladače běžící na pozadí.   |
| `time.sleep(time)` | Není součástí Selenia ale používáme ho často na zpomalení vykonávání abychonm viděli co se právě deje |

## 4. Simulace interakce uživatele
Zde nejdřív načteme Google stránku, pak potvrdíme cookies a následně hledáme na googlu "Python OOP tutorial".

```python
import time

from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager


# Initialize the service
service = Service(ChromeDriverManager().install())

options = Options()

# Start the session
driver = webdriver.Chrome(options=options)

# set max wait
driver.implicitly_wait(30)

driver.get("https://www.google.com")

try:
    agree_button = driver.find_element(By.ID, "L2AGLb") # click cookies confirmation
    agree_button.click()
except Exception:
    print("Consent overlay not found, proceeding.")

search_box = driver.find_element(By.NAME, "q") # Google uses name 'q' for search field

search_box.send_keys("Python OOP tutorial") # fill search field
search_box.send_keys(Keys.ENTER) # click ENTER

print(f"Page loaded: {driver.title}")

# wait to see output
time.sleep(3)

# terminate session
driver.quit()
```
